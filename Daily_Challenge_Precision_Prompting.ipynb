# Prompt Engineering Exercise: Remote Access Security Microlearning
### Learning & Communications Team — Internal Newsletter Content

**Goal:** Design, test, and refine a prompt that performs four control tasks on a real policy paragraph:
1. Tone, structure, format, and length control
2. Evaluation of output quality
3. Hallucination detection and mitigation
4. Paraphrasing vs. quoting techniques

**Original Policy Text (Input Dataset):**

> Employees must ensure that all remote access to internal systems is established via the approved secure VPN. Under no circumstances should unsecured connections or personal devices lacking endpoint protection be used to access proprietary data or sensitive communications.


## Step 1: Craft a Prompt That Controls Output Tone, Format, and Length

**Requirements:**
- Friendly and clear tone
- Bullet points
- Paraphrased (no direct quotes)
- Under 75 words


```python
original_text = (
    "Employees must ensure that all remote access to internal systems is "
    "established via the approved secure VPN. Under no circumstances should "
    "unsecured connections or personal devices lacking endpoint protection "
    "be used to access proprietary data or sensitive communications."
)

prompt_v1 = f"""You are an internal communications specialist writing for a company newsletter.
Rewrite the text below in a friendly, clear tone for non-technical employees.
Paraphrase the content — do not use direct quotes.
Organize it into 3–4 bullet points.
Keep the total response under 75 words.

Text: "{original_text}"
"""

print(prompt_v1)
```

**Generated Output (from GPT-4 / Gemini / model of choice):**

**🔒 Staying Safe with Remote Access**
- Always connect to work systems through the company's approved VPN.
- Avoid using public or unsecured networks to access work data.
- Only use devices that have proper security software installed.
- Never open sensitive files or messages on personal, unprotected devices.

*(~42 words)*


## Step 2: Evaluate the Output

| Criteria | Assessment |
|---|---|
| **Relevance** | Mostly on-target — covers VPN use, unsecured connections, and endpoint protection. |
| **Clarity** | Yes, plain language, no jargon like "endpoint protection" or "proprietary data." |
| **Structure** | Clean bullet points, scannable for a newsletter. |
| **Tone** | Friendly, appropriate for internal comms. |
| **Length** | Under 75 words ✅ |
| **Factual Accuracy** | ⚠️ Partial hallucination — "public or unsecured networks" is an inference the model added; the original only says "unsecured connections," not specifically public Wi-Fi. This is a plausible-sounding but unstated detail. |

**Finding:** This is a subtle hallucination — not a fabricated fact, but an unstated example presented as if it came from the source. This risks setting a slightly wrong expectation (e.g., an employee might assume a private-but-unapproved network is fine).


## Step 3: Detect and Mitigate Hallucinations

**Revised Prompt** — adds an explicit constraint to prevent the model from introducing unstated examples or technologies.


```python
prompt_v2 = f"""You are an internal communications specialist writing for a company newsletter.
Rewrite the text below in a friendly, clear tone for non-technical employees.
Paraphrase only the content explicitly stated in the input — do not add new examples,
technologies, or recommendations not present in the original text.
Organize it into 3–4 bullet points.
Keep the total response under 75 words.

Text: "{original_text}"
"""

print(prompt_v2)
```

**Revised Output:**

**🔒 Staying Safe with Remote Access**
- Always use the approved secure VPN to connect to internal systems remotely.
- Don't use unsecured connections to access company data.
- Avoid personal devices that lack proper security protection.
- Keep proprietary data and sensitive communications off unprotected setups.

This version sticks strictly to what's in the source — no invented examples like "public Wi-Fi."


## Step 4: Paraphrasing Deep Dive (Intern Audience)

**Requirements:**
- Plain language, short phrases
- No more than 4 bullet points
- No corporate or legal jargon
- Supportive, informative tone


```python
prompt_v3 = f"""Paraphrase the policy text below for a junior intern audience.
Use plain, simple language with short phrases — no corporate or legal jargon.
Keep a supportive, encouraging tone.
Use no more than 4 bullet points.

Text: "{original_text}"
"""

print(prompt_v3)
```

**Generated Output:**

**👋 A Quick Tip on Working Remotely**
- Use the company VPN when connecting from outside the office.
- Skip any unsecured or unofficial connections.
- Only use devices that have the right security tools installed.
- This helps keep company info and messages safe — thanks for helping protect it!

**Note:** Jargon like "proprietary data" and "endpoint protection" was replaced with plain terms like "company info" and "the right security tools" — same meaning, more accessible for a non-expert audience.


## Step 5: Quote Extraction Variant

**Prompt:** Extract one direct quote that best captures the core security policy.


```python
prompt_v4 = f"""Extract the single sentence from the text below that most directly and
completely states the core security requirement. Return it as a direct quote, unmodified.

Text: "{original_text}"
"""

print(prompt_v4)
```

**Output:**

> "Employees must ensure that all remote access to internal systems is established via the approved secure VPN."

### Reflection Questions

**In what kind of internal communication would quoting be more appropriate than paraphrasing?**

In formal or compliance-sensitive communications — official policy memos, legal notices, audit documentation, or anywhere an employee or auditor might need to point back to the *exact* wording that was communicated. Quoting preserves precision and creates a clear paper trail to the source policy.

**When might quoting pose a risk?**

When the audience is non-technical or time-pressed. The original phrasing ("endpoint protection," "proprietary data") can be confusing or easy to skim past. Quoting also can't adapt tone, so it risks sounding cold or overly formal for a friendly newsletter context — reducing engagement and retention of the actual safety message.


## Summary
