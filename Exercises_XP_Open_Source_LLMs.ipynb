## Exercise 1: Open Source Levels Reflection
**Use:** Fully Open, Weights Released, Architecture Only; model components; fine-tuning/domain adaptation; healthcare compliance.

**Step-by-step**
1) Gather definitions for the three openness levels.
2) Identify key characteristics (what is open; what you can/cannot do).
3) Compare side by side in a table (what's open vs. impact).
4) Write a 3-5 sentence comparative paragraph.
5) Answer the healthcare prompt (1-2 sentences) about retraining on clinical data.

**Deliverables:** paragraph + healthcare answer.


```python
open_source_levels = {
    "Fully Open": {
        "definition": (
            "Every component of the model pipeline is released: pretraining data (or a fully "
            "documented data recipe), training and evaluation code, model weights, hyperparameters, "
            "and technical reports, usually under a permissive license (e.g., Apache 2.0). "
            "Examples: OLMo (AI2), Pythia (EleutherAI), BLOOM (BigScience)."
        ),
        "what_is_open": [
            "Training data / documented data mixture",
            "Training + evaluation code and hyperparameters",
            "Final and intermediate model weights",
            "Full technical documentation / paper",
        ],
        "what_you_can_do": [
            "Reproduce training from scratch and verify claims",
            "Audit training data for bias, PII, or copyrighted content",
            "Retrain or fine-tune on new domains (including sensitive data) with full traceability",
            "Redistribute modified versions under the same permissive terms",
        ],
        "what_you_cannot_do": [
            "Nothing is legally/technically blocked, but compute cost to retrain from scratch is still very high",
            "Reproducing frontier-scale fully open models still requires large infrastructure",
        ],
    },
    "Weights Released": {
        "definition": (
            "Only the trained model artifact is released: final weights, tokenizer, and inference "
            "code, plus a model card. The training data and full training pipeline are not disclosed. "
            "Examples: Llama 2 / Llama 3 (Meta), Mistral-7B, Qwen2.5."
        ),
        "what_is_open": [
            "Final model weights (often in multiple precisions/quantizations)",
            "Inference code and tokenizer",
            "Model card with high-level training description",
            "Sometimes fine-tuning/LoRA scripts",
        ],
        "what_you_can_do": [
            "Run inference locally or in your own cloud",
            "Fine-tune or LoRA-adapt on top of the released weights",
            "Deploy commercially, subject to the specific license terms",
        ],
        "what_you_cannot_do": [
            "Reproduce the original training run from scratch",
            "Independently audit the exact pretraining data for compliance (e.g., verify no PII/copyrighted clinical text was included)",
            "Fully explain or certify the provenance of the base model's behavior",
        ],
    },
    "Architecture Only": {
        "definition": (
            "Only the model design is public - a paper or code skeleton describing the architecture "
            "and training method - with no weights and no data released. Historically this described "
            "releases like the original GPT-3 paper before any open weights existed."
        ),
        "what_is_open": [
            "Architecture design and paper",
            "Sometimes a minimal, untrained reference implementation",
        ],
        "what_you_can_do": [
            "Reimplement and train your own model from scratch if you have the data and compute",
            "Learn from the design choices and apply them to your own models",
        ],
        "what_you_cannot_do": [
            "Use the original trained model at all",
            "Fine-tune or query the original model's capabilities",
            "Get any usable model without first sourcing data and doing your own training run",
        ],
    },
}
open_source_levels
```

```python
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Data, code, weights, docs | Can retrain from scratch, fully audit, and adapt to any domain (including sensitive data) with traceability |
| Weights Released | Final weights + inference code | Can fine-tune/adapt starting from the released weights, but cannot reproduce or audit the original training run |
| Architecture Only | Design/paper only | Must source your own data and compute to train a usable model from scratch |"""
print(comparison_table)
```

```python
comparative_paragraph = """The three openness levels differ mainly in how much of the training pipeline - not just the
final model - is exposed. Fully Open models like OLMo or Pythia give you data, code, and weights, so you
can reproduce, audit, and safely adapt them to sensitive domains because you know exactly what went into
them. Weights Released models like Llama 2/3 or Mistral are far more common and are usually good enough
for production use, but because the training data and pipeline stay private, you can fine-tune the model
without ever being able to fully verify what it originally learned from. Architecture Only releases sit at
the other extreme: they document a design, not a usable artifact, so they are mainly useful for researchers
who plan to train their own model rather than teams looking to deploy quickly. In practice, most companies
choose Weights Released models for speed and performance, and reserve Fully Open models for cases where
auditability or regulatory traceability matters more than raw capability."""
print(comparative_paragraph)
```

```python
healthcare_prompt_answer = """Fine-tuning a Weights-Released model on private clinical data does not make the base model itself
HIPAA-compliant: you still cannot verify what the original pretraining data contained, so compliance has to
come from controlling your own fine-tuning data pipeline, access controls, and audit trail rather than from
trusting the vendor's original training process."""
print(healthcare_prompt_answer)
```
